```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Cargar y limpiar datos 
ruta = "../data/Superstore_Orders.csv"
df = pd.read_csv(ruta)

df['Sales'] = df['Sales'].str.replace(',', '.').astype(float)
df['Profit'] = df['Profit'].str.replace(',', '.').astype(float)
df['Discount'] = df['Discount'].str.replace(',', '.').astype(float)

print("Datos listos. Filas:", len(df))
```

```python
# Agrupar y sumar profit
segmento_profit = df.groupby('Segment')['Profit'].sum().sort_values(ascending=False)

print ("Ganancias totales por segmento:")
print(segmento_profit)  

# grafico
plt.figure(figsize=(8, 5))
sns.barplot(x=segmento_profit.index, y=segmento_profit.values,hue=segmento_profit.index, palette='viridis',legend=False)
plt.title('Ganancias Totales por Segmento de Clientes', fontsize=14)
plt.ylabel("Ganancias Totales")
plt.xlabel("Segmento")

# plt.savefig('../outputs/grafico_segmentos.png', bbox_inches='tight')

plt.show()
```

```python
# Top 10 clientes por ganancias
top_clientes = df.groupby('Customer Name')['Profit'].sum().sort_values(ascending=False).head(10)

print("Top 10 clientes por ganancias:")
print(top_clientes)

# grafico
plt.figure(figsize=(10, 6))
sns.barplot(x=top_clientes.values, y=top_clientes.index, hue=top_clientes.index, palette='magma',legend=False)
plt.title('Top 10 Clientes más Rentables', fontsize=14)
plt.xlabel('Ganancia Total (USD)')
plt.ylabel('Cliente')

#plt.savefig('../outputs/top_clientes.png', bbox_inches='tight')

plt.show()
```
